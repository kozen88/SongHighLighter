L'idea è quella di effettuare delle modifiche che ci permettano di trattare al meglio i dati e l'esporazione per permetterci di analizzare e rispondere a delle domande di business su questi dati. La prima cosa che notiamo consiste nel fatto che la maggiorparte delle colonne del dataframe costituiscono caratteristiche della musica inerenti alla canzone come proprietà fisiche delle onde sonore, per tale ragione nell'ottica di voler cnofrontare queste variabili assieme per cogliere aspetti più di alto livello nell'informazione che possiamo trarre da tali dati conviene apportare una elaborazione che permetta di portare ogni variabile numerica continua in uno spazio di range [0, 1] che ci permettera di definire uno spazio multidimensionale normalizzato nel quale confrontare equamente tali caratteristiche e definire degli assetti che qualifichino in qualche modo il suono e ci facciamo percepire alcune cambiamenti dovuti al genere all'anno o ad altri aspetti o fattori interagenti con tali caratteristiche.

A tal fine andremo a standardizzare le caratteristiche del suono che ancora non lo sono come:

- popularity: ovvero il punteggio della popolarità assegnato che va da 0 a 100 lo riportiamo a [0, 1]
- loudness: la potenza acustica del suona la riporteremo in uno standard [0, 1] per capirne il cambiamento

mentre per queste:

- duration_ms: ovvero al adurata della canzone la mettiamo in un formato più leggibile minuti e secondi
- tempo: il ritmo della canzone lo lasciamoo cosi come è caratteristico e semplice da capire rispetto alle altre caratteristiche acustiche

#### Normalizzazione e Raffinamento delle Metriche

Per garantire la confrontabilità tra le diverse dimensioni del dataset senza perdere il valore semantico delle informazioni fisiche, è stata adottata una strategia di scaling differenziata:
1. Normalizzazione $[0, 1]$: Applicata a loudness e popularity. La prima è stata trasformata per eliminare la scala logaritmica negativa dei decibel, facilitando la sovrapposizione grafica con variabili come energy. La seconda è stata riscalata per allinearsi al range degli altri indici di performance di Spotify.
2. Riformattazione Temporale: La variabile duration_ms è stata convertita nel formato leggibile minuti.secondi. Questa scelta di Data Design permette di mantenere l'interpretabilità intuitiva della durata di un brano, evitando l'astrazione derivante da una normalizzazione standard.
3. Preservazione del Dato Originale: Il tempo (BPM) è stato mantenuto nella sua scala originale, in quanto il valore numerico assoluto rappresenta un metadato culturale e tecnico fondamentale per l'identificazione del genere musicale.

```python
# Convertiamo i millisecondi in secondi totali
# Usiamo la divisione intera // per scartare i decimali e ottenere un intero
df['duration_seconds'] = (df['duration_ms'] // 1000).astype(int)

# Verifichiamo che non ci siano valori pari a 0 (per sicurezza nelle analisi)
df = df[df['duration_seconds'] > 0]
```

```python
# check sulla trasformazione temporale millisecondi a formato minuti: secondi
df[['duration_seconds','duration_ms' ]]
```

## Feature Engineering: Decodifica e Arricchimento della Tonalità

Per elevare la capacità comunicativa delle visualizzazioni e permettere un'analisi granulare del "mood" musicale, è stato necessario intervenire sulle variabili key (chiave musicale) e mode (modalità), originariamente espresse in formato numerico.

Logica di Elaborazione
Sebbene i valori interi (0-11 per la chiave, 0-1 per la modalità) siano ottimali per l'ordinamento sequenziale, essi risultano di difficile interpretazione per l'utente finale. Abbiamo quindi implementato una pipeline di trasformazione che genera tre nuovi livelli di informazione:

- key_name: Conversione della Pitch Class numerica nella nomenclatura anglosassone standard (C, C#, D, ecc.).

- mode_name: Esplicitazione della modalità in formato testuale (Major o Minor).

- tonality: Una variabile sintetica ottenuta dalla concatenazione delle precedenti (es. "A Minor", "C# Major").

### Legenda Variabile `key` (Pitch Class Notation)

| Valore | Nota (Anglosassone) | Nota (Italiana) |
| :---: | :---: | :--- |
| **0** | **C** | Do |
| **1** | **C# / Db** | Do# / Reb |
| **2** | **D** | Re |
| **3** | **D# / Eb** | Re# / Mib |
| **4** | **E** | Mi |
| **5** | **F** | Fa |
| **6** | **F# / Gb** | Fa# / Solb |
| **7** | **G** | Sol |
| **8** | **G# / Ab** | Sol# / Lab |
| **9** | **A** | La |
| **10** | **A# / Bb** | La# / Sib |
| **11** | **B** | Si |

> **Nota Metodologica:** La colonna `mode` definisce se la tonalità è Maggiore (1) o Minore (0). In musica, la combinazione di `key` e `mode` determina l'armonia complessiva del brano.

```python
# mapping for key e mode
key_map = {
    0: 'C', 1: 'C#', 2: 'D', 3: 'D#', 4: 'E', 5: 'F', 
    6: 'F#', 7: 'G', 8: 'G#', 9: 'A', 10: 'A#', 11: 'B'
}

mode_map = {1: 'Major', 0: 'Minor'}

# new columns string for key and mode
df['key_name'] = df['key'].map(key_map)
df['mode_name'] = df['mode'].map(mode_map)

# tonality as union of key and mode
df['tonality'] = df['key_name'] + " " + df['mode_name']
```

```python
import unicodedata

def final_clean(text):
    if not isinstance(text, str): return text
    
    
    t = text.lower().strip()
    #  Rimuoviamo il "the " all'inizio (es: "the script" -> "script")
    if t.startswith("the "):
        t = t[4:]
        
    # Rimuoviamo l'asterisco (es: "*nsync" -> "nsync")
    t = t.replace('*', '')
    
    # Puliamo Axwell /\ Ingrosso e simili (simboli grafici)
    t = t.replace('/\\', ' ').replace('/', ' ').replace('\\', ' ')
    
    # De-accentazione (es: Beyoncé -> Beyonce, Tiësto -> Tiesto)
    t = "".join(c for c in unicodedata.normalize('NFKD', t) if not unicodedata.combining(c))
    
    # Rimuoviamo doppi spazi bianchi
    t = " ".join(t.split())
    
    return t
```

```python
# Applichiamo tutto alla variabile di appoggio
df_enc['artist_preview'] = df_enc['artist'].apply(final_clean)

# VERIFICA FINALE: Ci sono stati accorpamenti?
original_count = len(df_enc['artist'].str.lower().unique())
cleaned_count = len(df_enc['artist_preview'].unique())

print(f"Artisti unici prima: {original_count}")
print(f"Artisti unici dopo la pulizia: {cleaned_count}")
print(f"Abbiamo accorpato {original_count - cleaned_count} varianti dello stesso nome!")
```
